## Descripción
Este trabajo utiliza un conjunto de datos sintético de 200 ventas de una tienda minorista. Cada fila representa una venta e incluye fecha, producto, categoría, cantidad, precio, ciudad, método de pago, descuento y total. Los datos son simulados con fines académicos; no representan transacciones reales.

## Objetivos

- Familiarizarse con Python, Pandas, NumPy y Matplotlib.
- Explorar la estructura, calidad y distribución de los datos.
- Obtener indicadores y visualizaciones para interpretar las ventas.

```python
# Importación de bibliotecas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
plt.rcParams['figure.figsize'] = (9, 5)
print('Bibliotecas cargadas correctamente.')
```

## 1. Carga del conjunto de datos
El archivo CSV se encuentra en la misma carpeta que este notebook.

```python
df = pd.read_csv('dataset_ventas.csv', parse_dates=['fecha'])
print('Datos cargados correctamente.')
df.head()
```

## 2. Exploración inicial
Se revisan las dimensiones, los nombres de columnas, los tipos de datos y un resumen estadístico.

```python
print('Filas y columnas:', df.shape)
print('\nInformación general:')
df.info()
print('\nResumen estadístico:')
display(df.describe(include='all').T)
```

## 3. Calidad de los datos
Se comprueban valores nulos y registros duplicados.

```python
print('Valores nulos por columna:')
display(df.isnull().sum().to_frame('valores_nulos'))
print('Filas duplicadas:', df.duplicated().sum())
print('Identificadores únicos:', df['id_venta'].nunique())
```

## 4. Análisis de ventas
Se calculan indicadores generales y agrupaciones por producto, categoría, ciudad y método de pago.

```python
ventas_totales = df['total'].sum()
promedio_venta = df['total'].mean()
unidades_vendidas = df['cantidad'].sum()
print(f'Cantidad de ventas: {len(df)}')
print(f'Ingresos totales: Bs {ventas_totales:,.2f}')
print(f'Promedio por venta: Bs {promedio_venta:,.2f}')
print(f'Unidades vendidas: {unidades_vendidas}')

print('\nIngresos por categoría:')
display(df.groupby('categoria')['total'].sum().sort_values(ascending=False).to_frame('ingresos_Bs'))
print('\nIngresos por ciudad:')
display(df.groupby('ciudad')['total'].sum().sort_values(ascending=False).to_frame('ingresos_Bs'))
print('\nProductos más vendidos:')
display(df.groupby('producto')['cantidad'].sum().sort_values(ascending=False).head(5).to_frame('unidades'))
```

## 5. Visualización de datos
Los siguientes gráficos permiten comparar ingresos y observar la evolución temporal.

```python
# Ingresos por categoría
ingresos_categoria = df.groupby('categoria')['total'].sum().sort_values(ascending=False)
ingresos_categoria.plot(kind='bar', title='Ingresos por categoría', ylabel='Ingresos (Bs)', xlabel='Categoría')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Cantidad de unidades vendidas por producto
unidades_producto = df.groupby('producto')['cantidad'].sum().sort_values(ascending=True)
unidades_producto.plot(kind='barh', title='Unidades vendidas por producto', xlabel='Unidades', ylabel='Producto')
plt.tight_layout()
plt.show()
```

```python
# Evolución mensual de ingresos
df['mes'] = df['fecha'].dt.to_period('M').astype(str)
ventas_mensuales = df.groupby('mes')['total'].sum()
ventas_mensuales.plot(kind='line', marker='o', title='Evolución mensual de ingresos', ylabel='Ingresos (Bs)', xlabel='Mes')
plt.xticks(rotation=45, ha='right')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
